## Metadata

### Candidates' data (df_applicants)
This dataset consisted of the following columns:
* `ID_INDIVIDUO`: Unique identifier of the applicant (those that were, at some point before or after the application, students at Nova IMS).
* `CD_CANDIDATO`: Anotheridentifier of the applicant (not unique for each applicant).
* `idcandidato`: Unique identifier of the applicant.
* `idgenero`: Gender of the applicant.
* `cd_pais`: Two-letter identifier of the applicant's country of origin (ISO 3166-1 alpha-2).
* `Localidade`: City/Place from where the applicant is from.
* `DGrauAcademicoPT`: Previous academic degree attained by the applicant.
* `DEstabEnsinoProvPT`: Previous HEI attended by the applicant.
* `DCursoProvPT`: Previous academic program attended by the applicant.
* `CursoConcluido`: Indicator of whether the applicant has already concluded the aforementioned program.
* `NotaOuMedia`: Grade of completion or current GPA (at the time of the application) of the aforementioned program.
* `DescEscala`: Grading scale on which the aforementioned program was/is evaluated.
* `nivelInglesCompreensao`: Applicant's comprehension level in English.
* `nivelInglesFala`: Applicant's oral level in English.
* `nivelInglesEscrita`: Applicant's writing level in English.
* `idCandidatura`: Unique identifier of the application.
* `datacandidaturafim`: Date and time on which the application was complete.
* `dprograma`: Academic degree to which the applicant is applying.
* `dcurso`: Program to which the applicant is applying.
* `DCursoOpcao`: Detail of the program to which the applicant is applying.
* `idEstadoCandidatura`: Identifier of the application status.
* `DEstadoPT`: Application status.

Each row represents a past or current academic degree from a certain applicant performing a specific application.

### Professional Activity (df_profactiv)
This dataset consists of the following columns:
* `idCandidato`: Unique identifier of the applicant.
* `DataInicio`: Starting date on the referred professional experience.
* `DataFim`: Finish date on the referred professional experience.
* `ActivProEntidadePatronal`: Employer of the applicant in the referred professional experience.
* `ActivProFuncao`: Name of the position occupied by the applicant in the referred professional experience.

Each row represents a past or current professional experience from an applicant.

### Students' data (df_students)
This dataset consisted of the following columns:
* `id_individuo`: Unique identifier of the student.
* `DT_MATRIC`: Date in which the student enrolled in the program.
* `dataConclusao`: Date of completion of the program.
* `NotaFinal`: Final grade of completion of the program.
* `situacaoFinal`: Student's enrollment status.
* `cd_lectivo`: School year on which the course was taken.
* `cd_curso`: Unique identifier of the program the student is enrolled in.
* `nm_curso_pt`: Name of the program the student is enrolled in.
* `cd_discip`: Unique identifier of the course.
* `ds_discip_pt`: Name of the course
* `notaFinalDisciplina`: Final grade of the course.

Each row represents a student taking a specific course.

```python
df_applicants_explore.groupby("CD_CANDIDATO")["idCandidatura"].nunique().sort_values(ascending = False)
```

Our idea does not confirm, as there are up to 6 different values of `idCandidatura` for the same `CD_CANDIDATO`. On the other hand, each `idCandidatura` is always associated with the same `CD_CANDIDATO`.

The observations above have the same `CD_CANDIDATO`, but they almost surely don't belong to the same person, since we have different genders, countries, and locations for these people. Therefore, and since it doesn't provide any link to relate with the remaining tables at our disposal, we will __discard the `CD_CANDIDATO` variable__, and proceed with the remaining 3 "ID" variables. Before taking more conclusions, let's check that each application (`idCandidatura`) is always associated with the same `idcandidato` (and, consequently, the same `ID_INDIVIDUO`).

```python
df_applicants_explore.drop(columns = {"CD_CANDIDATO"}, inplace = True)
```

```python
df_applicants_explore.groupby("idCandidatura")["idcandidato"].nunique().sort_values(ascending = False)
```

Great. Now, we have seen that the same `idCandidatura` can appear more than once in our data. Let's see how common this is.

```python
df_applicants_explore["idCandidatura"].value_counts()
```

```python
print("There are",
      len(df_applicants_explore["idCandidatura"].value_counts()[lambda x: x > 1]),
      "values of 'idCandidatura' that appear more than once in our data.")
print("Conversely, there are",
      len(df_applicants_explore["idCandidatura"].unique()),
      "unique applications in our dataset.")
```

From the analysis made up to this point, we can now summarize our interpretation of what the "ID" variables mean:
* `ID_INDIVIDUO` and `idcandidato` - they represent a person, the applicant/student. Each student can make more than one application, but an application can only be made by one person.
* `idCandidatura` - represents an application to a certain program, in a certain moment of time. For instance, if the same person applies to the same masters' program in two different school years, they will have two different values for `idCandidatura`.

This is where we want to base our predictions. __We want each row to be an application__, not a part of an application and even less a person. Therefore, we have to transform our dataset to reflect this. At the moment, each row represents a past or current academic degree from a certain applicant performing a specific application. Therefore, the only variables that can change are `DGrauAcademicoPT`, `DEstabEnsinoProvPT`, `DCursoProvPT`, `CursoConcluido`, `NotaOuMedia`, and `DescEscala`.

```python
cols_to_pivot = ["DGrauAcademicoPT", "DEstabEnsinoProvPT", "DCursoProvPT", "CursoConcluido", "NotaOuMedia", "DescEscala"]

df_applicants_explore = tf.pivot_columns(df_applicants_explore, cols_to_pivot, "idCandidatura")
df_applicants_explore.head()
```

Great! We transformed our data to reflect the analysis we intend to conduct.

The last step before moving to data exploration in itself is to __limit the scope of our work to applicants to the masters' and postgraduates' programs__, since we were already able to identify through the `dprograma` variable that there are also applications to other education levels other than these two.

Note: This and other transformations will later be implemented in the original df_applicants dataframe. For now, we are just interested in adapting data to our exploration interests.

```python
df_applicants_explore["dprograma"].value_counts()
```

```python
programs_to_keep = ["Mestrado", "Pós-Graduação", "Master", " Postgraduate", "Post-Graduate", " Pós-Graduação"]

df_applicants_explore = df_applicants_explore[df_applicants_explore["dprograma"].isin(programs_to_keep)]
```

Let's now analyse the variables related to the dates of the professional experiences (`DataInicio` and `DataFim`). First, we can assess the distribution of the number of years of experience and how their averages vary accross the different admission status.

Note: When we arrive to preprocessing, we will use more precise statistical tests in order to assess feature importance. For now, we are attempting to get a glimpse of the data, identify some hints and potential patterns to base our actions later on.

```python
start_date_columns = ["DataInicio_1", "DataInicio_2", "DataInicio_3", "DataInicio_4", "DataInicio_5", "DataInicio_6"]

df_applicants_profactiv_explore["datacandidaturafim_year"] = pd.to_numeric(
    df_applicants_profactiv_explore["datacandidaturafim_year"])

df_applicants_profactiv_explore["Years_of_Experience"] = (
    df_applicants_profactiv_explore["datacandidaturafim_year"] - df_applicants_profactiv_explore[start_date_columns].min(axis=1).dt.year
).fillna(0).clip(lower = 0).astype(int)
```

```python
tf.summary_statistics(df_applicants_profactiv_explore, "DEstadoPT", "Years_of_Experience")
```

Interestingly, even though the difference may not be strongly significative (considering that standard deviations are relatively high), it seems that applicants that were admitted ("Admitido" or "Desistiu após seleção") have, on average and on median, less years of professional experience than those who were rejected ("Não Admitido").

It could also be interesting to see how the length of the longest professional experience relates to the admission status.

```python
end_date_columns = ["DataFim_1", "DataFim_2", "DataFim_3", "DataFim_4", "DataFim_5", "DataFim_6"]

durations = []
for inicio_col, fim_col in zip(start_date_columns, end_date_columns):
    start_year = df_applicants_profactiv_explore[inicio_col].dt.year
    end_year = df_applicants_profactiv_explore[fim_col].dt.year.fillna(df_applicants_profactiv_explore["datacandidaturafim_year"])
    duration = (end_year - start_year).fillna(0).clip(lower = 0)
    durations.append(duration)

df_applicants_profactiv_explore["Years_of_Longest_Prof_Experience"] = pd.concat(durations, axis = 1).max(axis = 1)
```

```python
tf.summary_statistics(df_applicants_profactiv_explore, "DEstadoPT", "Years_of_Longest_Prof_Experience")
```
